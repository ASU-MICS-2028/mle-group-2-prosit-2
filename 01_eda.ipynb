```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
train = pd.read_csv('datasets/Train.csv')
test = pd.read_csv('datasets/Test.csv')
print('Train:', train.shape)
print('Test:', test.shape)
```

```python
print(train['target'].value_counts())
print('Default rate:', round(train['target'].mean() * 100, 2), '%')
```

```python
print(train.isna().sum())
```

```python
train['extra_pct'] = (train['Total_Amount_to_Repay'] / train['Total_Amount'] - 1) * 100
print(train.groupby('target')['extra_pct'].median())
```

```python
rate = train.groupby('New_versus_Repeat')['target'].mean() * 100
print(rate)

rate.plot(kind='bar')
plt.ylabel('Default rate (%)')
plt.title('Default rate: new vs repeat borrowers')
plt.show()
```

```python
train['disbursement_date'] = pd.to_datetime(train['disbursement_date'])
train['year'] = train['disbursement_date'].dt.year

by_year = train.groupby('year')['target'].agg(['count', 'mean'])
by_year['mean'] = by_year['mean'] * 100
print(by_year)
```
